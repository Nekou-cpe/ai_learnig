###importing library
1-np

```python
import numpy as np
print(np.array([21,3,4,5,33]))
print(np.array([21,3,4,5,33],ndmin=5))
print(np.array([[21,3,4,5,33],[23,44,55,66,7]]))
print(np.array([[21,3,4,5,33],[23,44,55,66,7]],dtype=float))
print(np.array([[21,3,4,5,33],[23,44,55,66,7]],dtype=complex))
```

zero/one/full matrix    

```python
zeroarr=np.zeros((2,2),dtype=int)
print(zeroarr)
```

```python
onearr=np.ones((2,2),dtype=int)
print(onearr)
```

```python
fullarr=np.full((3,3),10)
print(fullarr)
```

vector random full what you want?

```python
vector=np.random.randint(10,size=5)
print(vector)
emptyarr=np.empty_like(vector)
print(emptyarr)
onelikearr=np.ones_like(vector)
print(onelikearr)
zerolikearr=np.zeros_like(vector)
print(zerolikearr)
fulllike=np.full_like(vector,10)
print(fulllike)
```

transose matrix

```python
v1=np.random.randint(12,size=(3,4))
print(v1)
print(v1.T)
```

tril or triu


```python
y=np.tril(v1)
print(y)
print(np.tril(v1,-1))
x=np.triu(v1)
print(x)
print(np.triu(v1,+1))

print(x+y)#moteqaren matrix

```
